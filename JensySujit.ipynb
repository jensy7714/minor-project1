1. What is the difference between 'is' and '==' operator?

**Answer :** "==" operator checks whether two variables have the same value.

"is" operator checks whether two variables refer to same object in memory.

```python
#Example
a = ["apple", 10, "basket" ]
b = ["apple", 10, "basket" ]

print(a == b)
print(a is b)
```

5. **Without using the word "hash" :**

A dictionary can find a value quickly using its key.But a list may need to check each element one by one to find a particular value. So, dictionary lookup take O(1) time on average,  while list searching takes O(n) time.

**Using the word "hash" :**

A dictionary uses a hash function to find the storage position of a key. This makes searching faster, taking O(1) time on average. A list checks elements one by one, so  it takes O(n) time.

7. **Difference between list.append(x), list.extend(x), and list.insert(i,x)**

    list.append() - adds an item at the end.

```python
a = [1,2,3]
a.append([4,5])
print(a)
```

    list.extend() - adds multiple items separately.
    

```python
a = [1,2,3]
a.extend([4,5])
print(a)
```

    list.insert() - Adds an item at specific position.

```python
a = [1,2,3]
a.insert(2,[4,5])
print(a)
```
